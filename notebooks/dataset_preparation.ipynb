
# Task 1: Dataset Selection and Description

## About the Dataset

For this project, we selected the Breast Cancer Wisconsin (Diagnostic) dataset.

This dataset was downloaded from kaggle. It contains data collected from breast mass samples using Fine Needle Aspirate (FNA) images.

The main purpose of this dataset is to identify whether a breast tumor is benign or malignant.

The dataset has 569 rows and 32 columns.

## Features of the Dataset

The dataset contains 30 numerical features. These features describe different characteristics of cell nuclei.

The main features are:

1. Radius
2. Texture
3. Perimeter
4. Area
5. Smoothness
6. Compactness
7. Concavity
8. Concave Points
9. Symmetry
10. Fractal Dimension

Each feature has three measurements: Mean, SE, and Worst.

## Target Column

The target column is called diagnosis.

It has two values:

- B = Benign (not cancerous)
- M = Malignant (cancerous)

We will use this column to classify the tumor.



# Task 3: Data Cleaning and Preparation

Before using the dataset, we performed some basic cleaning steps.

- First, we removed the ID column because it is not needed for prediction.
- We also removed the Unnamed: 33 column because it does not contain useful data.
- We checked the dataset for missing values and duplicate records.
- We changed the diagnosis values into numbers. B was changed to 0 and M was changed to 1.
- Then, we separated the dataset into X and y.

X contains the input features, while y contains the target values.

These steps helped us prepare the dataset for machine learning.



# Task 4: Real-World Applications of kNN

## 1. Healthcare

kNN can be used in healthcare to help classify diseases using patient data.

For example, it can help classify breast tumors as benign or malignant by comparing their features with similar samples.

It checks the nearest data points and predicts the class based on the majority of nearby samples.

This can help doctors when studying medical data and making decisions.

## 2. Customer Classification

kNN can also be used in business to classify customers.

Companies can use information such as age, income, shopping habits, and purchase history.
